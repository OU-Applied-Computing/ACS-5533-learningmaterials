```python
import numpy as np
import pandas as pd
from sklearn.linear_model import Perceptron
```

#### Initialize dataset representing AND operation. 

```python
data = {
    'x1': [0, 0, 1, 1],
    'x2': [0, 1, 0, 1],
    'class': [0, 0, 0, 1]
}
df = pd.DataFrame(data)
print(df)
```

```python
X = df[["x1", "x2"]]
y = df[["class"]]
```

#### Initialize Perceptron
- tol: The stopping criterion. Training halts when loss drops below this value.
- eta0: Learning rate.
- max_iter: maximum number of epochs. 

```python
classifier = Perceptron(tol = 0.001, eta0 = 0.001, max_iter = 100)
classifier.fit(X, np.ravel(y))
```

#### Test

```python
y_pred = classifier.predict(pd.DataFrame({'x1': [0], 'x2': [0]}))
print(y_pred)
```

```python
y_pred = classifier.predict(pd.DataFrame({'x1': [0], 'x2': [1]}))
print(y_pred)
```

```python
y_pred = classifier.predict(pd.DataFrame({'x1': [1], 'x2': [0]}))
print(y_pred)
```

```python
y_pred = classifier.predict(pd.DataFrame({'x1': [1], 'x2': [1]}))
print(y_pred)
```

#### Print the weights and the bias

```python
print(classifier.coef_, classifier.intercept_)
```
